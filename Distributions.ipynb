2. Generate 100 values in Excel using the continuous uniform distribution RAND() and plot a histogram. Describe the shape of the distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate 100 values from a continuous uniform distribution [0, 1) mimicking Excel's RAND()
rand_values = np.random.rand(100)

# Plotting the histogram
plt.figure(figsize=(8, 5))
plt.hist(rand_values, bins=10, edgecolor='black', color='skyblue', alpha=0.7)
plt.title('Histogram of 100 Uniformly Distributed Values (RAND())')
plt.xlabel('Value Range [0, 1)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Distribution Shape Description

- **Theoretical Shape**: Theoretically, a continuous uniform distribution is perfectly flat, meaning every value within the interval $[0, 1)$ has an equal probability density of occurring.
- **Observed Shape**: In this sample of 100 generated values, we observe some fluctuations (some bins are taller than others) due to random sampling variance. However, it generally spreads out across the entire interval without showing any distinct single peak (like a bell curve) or a strong skewness, indicative of a rectangular or uniform distribution shape.

4. What is the concept of standardization (z-score), and why is it important in data analysis? Explain the formula and how standardization transforms a dataset.


Answer -  Standardization is the process of transforming data into z-scores so that different variables can be compared on a common scale. The z-score is calculated using the formula (z=(x-\mu)/\sigma), where x is the observation, μ is the mean, and σ is the standard deviation. After standardization, the dataset has a mean of 0 and a standard deviation of 1. Positive z-scores indicate values above the mean, while negative z-scores indicate values below the mean. Standardization is important in data analysis because it removes differences in measurement scales, making variables easier to compare and improving the performance of many statistical and machine-learning methods.



5. What is Kurtosis and their type?

Answer - Kurtosis is a statistical measure used to describe the shape of a distribution, particularly the heaviness of its tails and the concentration of observations around the center. There are three types of kurtosis: Mesokurtic, Leptokurtic, and Platykurtic. A mesokurtic distribution has kurtosis of 3 and resembles a normal distribution. A leptokurtic distribution has kurtosis greater than 3, indicating heavier tails and more extreme values. A platykurtic distribution has kurtosis less than 3, indicating lighter tails and fewer extreme values. In Excel, excess kurtosis is commonly used, where a normal distribution has a kurtosis value of 0.

8. How does log transformation help in stabilizing variance and making data more normally distributed?

Answer - Log transformation is used to reduce skewness and stabilize the variance of a dataset. It is commonly performed using \(Y=\log(X)\). The transformation compresses large values more strongly than small values, reducing the influence of extreme observations. This can make the variance more consistent across different levels of the data. For right-skewed data, log transformation also reduces skewness and can make the distribution more approximately normal. Therefore, it is useful for improving the assumptions and performance of many statistical analyses.
