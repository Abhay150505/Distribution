Q1: Understanding Central Tendency (Easy) A bakery tracks the daily sales of muffins (in dozens) over a week: [10, 12, 11, 15, 14, 13, 12]. What is the most representative value of their weekly sales, and why?

Sol - The most representative value of the bakery's weekly muffin sales is 12 dozen. The mean is 12.43 dozen, the median is 12 dozen, and the mode is 12 dozen. Since both the median and mode are 12 and the daily sales are relatively close to this value, 12 dozen provides a simple and representative measure of the bakery's typical daily sales.

Q5: Dispersion Introduction (Medium) A student times how long it takes to finish a puzzle each day: [25, 30, 27, 35, 40]. What does the range tell us about the variation in the student’s puzzle-solving time?

Sol - The range measures the difference between the highest and lowest values in a dataset. For the puzzle-solving times, the maximum time is 40 minutes and the minimum time is 25 minutes. Therefore, Range = 40 − 25 = 15 minutes. This means that the student's puzzle-solving time varied by 15 minutes during the five days. A range of 15 minutes indicates that there is some variation in the student's performance from day to day.

Q6: Range in Action (Medium) A farmer records the weekly weight of harvested apples (kg): [100, 105, 98, 110, 120]. Find the range. How can this help the farmer in planning his packaging?

Sol - The range is calculated by subtracting the minimum value from the maximum value.
\[
\text{Range}=120-98=22\text{ kg}
\]
Therefore, the range is 22 kg. This means that the weekly apple harvest varies by 22 kg, with the lowest harvest being 98 kg and the highest being 120 kg. This information helps the farmer plan packaging materials and storage capacity. By knowing the range, the farmer can prepare enough boxes or bags to handle the maximum expected harvest.

Q9: Combining Measures (Hard) A family records their monthly electricity usage (in kWh): [400, 420, 390, 450, 410].
Find the mean and standard deviation. What do these values together tell you about the family’s energy use pattern?

Sol - The family’s monthly electricity usage is [400, 420, 390, 450, 410] kWh. The mean is calculated as (400 + 420 + 390 + 450 + 410) ÷ 5 = 414 kWh. Using the population standard deviation formula, the squared deviations from the mean are 196, 36, 576, 1296, and 16, giving a total of 2120. Therefore, the standard deviation is √(2120 ÷ 5) = √424 ≈ 20.59 kWh. Thus, the family’s average monthly electricity usage is 414 kWh, while the standard deviation of approximately 20.59 kWh indicates relatively low variation around the average. Together, these measures show that the family has a fairly consistent energy-use pattern, with monthly consumption generally staying close to 414 kWh.

Q10: Practical Application (Hard) A basketball player’s points in 8 games are recorded: [15, 18, 20, 22, 25, 17, 19, 21]. Find the mean, median, mode, range, and standard deviation. What insights can these measures provide about the player’s scoring performance?

Sol - The basketball player’s scores in 8 games are [15, 18, 20, 22, 25, 17, 19, 21]. The mean score is (15 + 18 + 20 + 22 + 25 + 17 + 19 + 21) ÷ 8 = 19.625 ≈ 19.63 points. After arranging the scores as 15, 17, 18, 19, 20, 21, 22, and 25, the median is (19 + 20) ÷ 2 = 19.5 points. There is no mode because every score occurs only once. The range is 25 − 15 = 10 points. Using the population standard deviation formula, the standard deviation is √(68 ÷ 8) = √8.5 ≈ 2.91 points. These measures show that the player scores approximately 20 points per game and has a fairly consistent scoring performance. The small difference between the mean and median and the relatively low standard deviation indicate that the scores are generally concentrated around the average, with no major inconsistency.
